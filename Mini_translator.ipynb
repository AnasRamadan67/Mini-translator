```python
from textblob import Word
```

```python
print("Hello, I'm Word Teacher 👩🏻‍🏫🖐🏻")
print("Type one word and I will analize it with NLP")
print("-----------------------------------------------")

while True:
  user_word=input("\n Type the word you want to analize or enter 'quit' if you finished:").lower()
  if user_word == 'quit':
    print(" Bye bye 👋🏻")
    break

  w = Word(user_word)

  corrected =w.spellcheck()[0][0]

  lemma = Word(corrected).lemmatize('v')

  print(f"\n🔎 NLP Analysis for '{user_word}':")
  print(f"✅ Corrected spelling: {corrected}")
  print(f"🌱 Lemma (base form): {lemma}")
```
